```python
name = input("Enter full name: ")
age = int(input("Enter age: "))
course = input("Enter course: ")
employed = input("Employed (yes/no): ")
stipend = float(input("Enter monthly stipend: "))

print("\nStudent Profile")
print(f"Name: {name}")
print(f"Age: {age}")
print(f"Course: {course}")
print(f"Employed: {employed}")
print(f"Stipend: {stipend}")

print("\nName Analysis")
print("First three letters:", name[:3])
print("Uppercase:", name.upper())

student = (name, age, course, employed, stipend)
print("\nStudent Tuple:", student)

marks = [80, 75, 90, 75, 60]
print("\nMarks:", marks)

marks.append(85)
print("After adding 85:", marks)

marks.remove(60)
print("After removing 60:", marks)

marks.sort()
print("Sorted marks:", marks)

marks = list(set(marks))
marks.sort()
print("Without duplicates:", marks)

subject_marks = {"Python": 85, "Excel": 78, "MySQL": 90}
print("\nSubject marks:", subject_marks)

subject_marks["Python"] = 92
subject_marks["Power BI"] = 72
print("After update:", subject_marks)

subjects = frozenset(subject_marks)
print("\nFrozenset:", subjects)

online = {"Ramees", "Anu", "Kiran"}
offline = {"Anu", "Rahul", "Sana"}
print("\nCommon students:", online & offline)
print("All students:", online | offline)
print("Only online students:", online - offline)

average = sum(subject_marks.values()) / len(subject_marks)

if average >= 90:
    grade = "A"
elif average >= 75:
    grade = "B"
elif average >= 50:
    grade = "C"
else:
    grade = "Fail"

print(f"\nAverage: {average:.2f}")
print(f"Grade: {grade}")

print("\nData Types")
print("name:", type(name))
print("age:", type(age))
print("stipend:", type(stipend))
print("student:", type(student))
print("marks:", type(marks))
print("subject_marks:", type(subject_marks))
print("subjects:", type(subjects))
print("online:", type(online))
print("average:", type(average))
```
